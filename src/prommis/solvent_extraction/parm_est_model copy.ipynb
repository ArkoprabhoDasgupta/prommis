```python
import pandas as pd
from idaes.core.solvers import get_solver
from pyomo.environ import ConcreteModel, RangeSet, Var, log10, minimize, Constraint
```

```python
df = pd.read_excel("data for parmest.xlsx", sheet_name = "Sheet2 (2)")
index = df.shape[0]
idx_list = RangeSet(0,index-1)
print(df)
```

```python
Element_list = ['Y','Ce','Nd','Sm','Gd','Dy','La','Pr']
```

```python
m = ConcreteModel()

m.pH = Var(idx_list)
m.dosage = Var(idx_list)
# m.logD_exp = Var(idx_list, Element_list)
m.logD_model = Var(idx_list, Element_list)
m.E_exp = Var(idx_list, Element_list)
m.E_model = Var(idx_list, Element_list, bounds=(0, 100))
m.D_model = Var(idx_list, Element_list, bounds=(0, 1e5))


for i in idx_list:
    m.pH[i].fix(df["pH"][i])
    m.dosage[i].fix(df["dosage"][i])
    for e in Element_list:
        # m.logD_exp[i,e].fix(df[f'logD {e}'][i])
        m.E_exp[i, e].fix(df[f"E {e}"][i])

variable_list = [1, 2]

# m.a = Var(variable_list, Element_list, initialize = 0.1)
# m.b = Var(variable_list, Element_list, initialize = 0.1)
m.a = Var(
    variable_list,
    Element_list,
    initialize=0.5,
    bounds=lambda m, i, e: (0, None) if i == 2 else (None, None),
)
m.b = Var(
    variable_list,
    Element_list,
    initialize=0.5,
    bounds=lambda m, i, e: (0, None) if i == 2 else (None, None),
)

# for e in Element_list:
#     m.b[2,e].set_lb(0)


@m.Constraint(idx_list, Element_list)
def logD_model_rule(m, i, e):
    if df.loc[i, f"w_{e}"] == 0:
        return Constraint.Skip
    else:
        return m.logD_model[i, e] == (m.a[1, e] + m.a[2, e] * m.dosage[i]) * m.pH[i] + (
            m.b[1, e] + m.b[2, e] * log10(m.dosage[i])
        )


@m.Constraint(idx_list, Element_list)
def D_model_rule(m, i, e):
    if df.loc[i, f"w_{e}"] == 0:
        return Constraint.Skip
    else:
        return 10 ** m.logD_model[i, e] == m.D_model[i, e]


@m.Constraint(idx_list, Element_list)
def E_model_rule(m, i, e):
    if df.loc[i, f"w_{e}"] == 0:
        return Constraint.Skip
    else:
        return m.E_model[i, e] * ((m.D_model[i, e] + 1)) == m.D_model[i, e] * 100


for e in Element_list:
    for i in idx_list:
        if df.loc[i, f"w_{e}"] == 0:
            m.E_model[i, e].fix(1e-7)
            m.D_model[i, e].fix(1e-7)
            m.logD_model[i, e].fix(0)


@m.Expression()
def SSE(m):
    return sum(
        ((m.E_exp[i, e] - m.E_model[i, e]) ** 2)*df.loc[i, f"w_{e}"]
        for i in idx_list
        for e in Element_list
        if df.loc[i, f"w_{e}"] != 0
    )/sum(df.loc[i, f"w_{e}"] for i in idx_list for e in Element_list)


@m.Objective(sense=minimize)
def objective_rule(m):
    return m.SSE


solver = get_solver("ipopt_v2")
solver.solve(m, tee=True)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score

for e in Element_list:
    plt.rcParams.update({
        "figure.max_open_warning": 0,
        "figure.dpi": 300,
        "axes.titlesize": 16,
        "axes.labelsize": 14,
        "axes.linewidth": 2,
        "lines.linewidth": 2,
        "lines.markersize": 8,
        "xtick.labelsize": 12,
        "ytick.labelsize": 12,
        "savefig.bbox": "tight",
        "legend.fontsize": "large",
    })
    filtered_indices = [i for i in idx_list if df.loc[i, f"w_{e}"]!=0]
    x = [m.E_exp[i, e]() for i in filtered_indices]
    y = [m.E_model[i, e]() for i in filtered_indices]
    plt.figure(dpi=300)
    plt.plot(x, x, label='y=x (ideal)', linestyle='-', color='black')
    red_points = []
    green_points = []
    blue_points = []
    for j, i in enumerate(filtered_indices):
        if 2 < df['dosage'][i] < 5:
            red_points.append(plt.scatter(x[j], y[j], color='red', label='5% DEHPA 10% TBP'))
        elif df['dosage'][i] < 2:
            green_points.append(plt.scatter(x[j], y[j], color='green', label='2% DEHPA 10% TBP'))
        elif 5 < df['dosage'][i] <= 25:
            blue_points.append(plt.scatter(x[j], y[j], color='blue', label='25% DEHPA 10% TBP'))
    plt.xlabel('experiment')
    plt.ylabel('model')
    plt.title(f"{e} E parity, $R^2$ = {r2_score(x, y):.4f}")
    plt.minorticks_on()
    plt.grid(which="major")
    plt.gca().set_axisbelow(True)
    
    
    # Build legend handles/labels manually to avoid duplicates
    handles = [plt.Line2D([], [], color='black', label='y=x (ideal)')]
    if red_points:
        handles.append(plt.Line2D([], [], color='red', marker='o', linestyle='None', label='5% DEHPA 10% TBP'))
    if green_points:
        handles.append(plt.Line2D([], [], color='green', marker='o', linestyle='None', label='2% DEHPA 10% TBP'))
    if blue_points:
        handles.append(plt.Line2D([], [], color='blue', marker='o', linestyle='None', label='25% DEHPA 10% TBP'))
    labels = [h.get_label() for h in handles]
    plt.legend(handles, labels)
    
    plt.show()
```
